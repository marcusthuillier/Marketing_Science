# Model 06 — World Cup Winner Propensity Model

**Discipline:** Marketing DS  
**Method:** Logistic regression + SHAP explainability  
**Question:** Which national teams have the profile of a World Cup winner — and how similar is the 2026 field to past champions?

---

Propensity models score leads by their similarity to historical converters. This model does the same for national football teams: it learns what a World Cup winner looks like across measurable pre-tournament features and scores every team in the field against that profile.

## 0. Imports & Setup

```python
%matplotlib inline
from run_model_06 import (
    load_data, compute_elo, build_wc_history,
    build_feature_matrix, train_model, compute_shap,
    score_2026, generate_plots,
)
print('Ready.')
```

## 1. Data Loading

```python
raw = load_data()
raw.head(3)
```

## 2. Elo Computation

```python
elo_df, elo_at, form_stats = compute_elo(raw)
elo_df.tail(3)
```

## 3. WC History

```python
wc_hist_teams, wc_hist_matches, wc_all_winners_norm = build_wc_history(raw)
```

## 4. Feature Matrix

```python
df, le = build_feature_matrix(elo_at, form_stats, wc_hist_teams, wc_hist_matches, wc_all_winners_norm)
df.head(5)
```

## 5. Model Training

```python
model, train, test = train_model(df)
```

## 6. SHAP Analysis

```python
shap_vals, X_tr_sc = compute_shap(model, train)
```

## 7. Score 2026

```python
df_2026 = score_2026(elo_at, form_stats, wc_hist_teams, wc_hist_matches, wc_all_winners_norm, le, model)
df_2026.head(10)
```

## 8. Visualizations

```python
generate_plots(test, df_2026, shap_vals, X_tr_sc, model, df, show=True)
```
